````python
# FINAL CODE

# Import libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd

# Read data files
fp = os.path.join('data', 'Spill_Incidents_20261102.csv')
spills = pd.read_csv(fp)

fp = os.path.join('data', 'tl_2026_us_county', 'tl_2026_us_county.shp')
counties = gpd.read_file(fp)

# # Changing space to underscore for columns in Spill dataframe
spills.columns = (spills.columns
                        .str.lower()
                        .str.replace(' ', '_'))

# Convert column names to lowercase in counties dataframe
counties.columns = (counties.columns
                            .str.lower())

# Converting date columns from object to datetime64
spills['spill_date'] = pd.to_datetime(spills['spill_date'])

# Gather NY counties using state geocode
ny_counties = counties[counties['statefp'] == '36']

# Filter for Petroleum spills and the entire year of 2023
petr_23 = spills[
    (spills['material_family'] == 'Petroleum') & 
    (spills['spill_date'] >= '01/01/2023') & 
    (spills['spill_date'] <= '12/31/2023')]

# Create data frame with spill number per country and set it to the index 
spills_per_county = petr_23.groupby('county')['spill_number'].nunique().reset_index()

# Rename the column to n_spills
spills_per_county = spills_per_county.rename(columns={'spill_number': 'n_spills'})

# Update St Lawrence to create matching county names in both dataframes
spills_per_county['county'] = spills_per_county['county'].replace('St Lawrence', 'St. Lawrence')

# Use NY counties using state geocode
ny_counties = counties[counties['statefp'] == '36']

# Prepare by cleaning data to merge the two dataframes
ny_counties = (ny_counties.rename(columns={'name': 'county'})
                            .drop(columns=['statefp',
                                           'countyfp',
                                           'countyns',
                                           'geoid',
                                           'geoidfq']))

# Reset index to match datasets before merging
ny_counties = ny_counties.reset_index(drop=True)

# Merge ny_counties as left to keep geometry
ny_spills = pd.merge(ny_counties,
                     spills_per_county,
                     how='left',
                     on='county')

# Getting spills for Onodaga reservation
ono_spills = spills_per_county.iat[35, 1]
ono_spills

# Getting spills for Tuscarora reservation
tusc_spills = spills_per_county.iat[58, 1]
tusc_spills

# Add both the reservation spills into the neighbor county
ny_spills.loc[ny_spills['county'] == 'Niagara', 'n_spills'] = ny_spills.loc[ny_spills['county'] == 'Niagara', 'n_spills'] + tusc_spills
ny_spills.loc[ny_spills['county'] == 'Onondaga', 'n_spills'] = ny_spills.loc[ny_spills['county'] == 'Onondaga', 'n_spills'] + ono_spills

# Create choropleth map 
fig, ax = plt.subplots(figsize=(8, 6))

ny_spills.plot(ax=ax,
               column='n_spills',
               cmap='RdPu_r',
                      legend=True,
               edgecolor="0.6",
               linewidth=0.5,
               legend_kwds={
                   "shrink": 0.7,
                   "label": "Number of Oil Spills",
                   "orientation": "horizontal",
                   "pad": 0.05})

            
# Add title and remove axis grid
ax.set_title("New York Oil Spills in 2023")
ax.set_axis_off();

fig.text(
    0.99,
    0.01,
    "Source: New York State Department of Environmental Conservation, accessed: Nov. 2, 2026",
    ha="right",
    va="bottom",
    fontsize=9,
    color="gray",
    style="italic")
````
